```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

def mean_2d(X, stride=2):
    h, w = X.shape
    
    out_h = h // stride
    out_w = w // stride
    
    result = np.zeros((out_h, out_w))
    
    for i in range(out_h):
        for j in range(out_w):
            block = X[
                i*stride:(i+1)*stride,
                j*stride:(j+1)*stride
            ]
            
            result[i, j] = np.mean(block)
    
    return result

    
X=np.arange(16).reshape((4,4))
print(f"X:\n{X}")
print(f"X:\n{mean_2d(X,stride=2)}")
```

```python
import numpy as np

def my_tri(n=3):
    A = np.zeros((n, n), dtype=bool)
    
    for i in range(n):
        A[i, :i+1] = True
    
    return A

print(my_tri(5))
```

```python
class Time:
    def __init__(self, hour, minute):
        self.minute = hour * 60 + minute

    def __str__(self):
        h = self.minute // 60
        m = self.minute % 60
        return f"{h} hours and {m} minutes"

    def __add__(self, other):
        total = self.minute + other.minute
        
        h = total // 60
        m = total % 60
        
        return Time(h, m)

    def __lt__(self, other):
        return self.minute < other.minute


t1 = Time(2, 10)
t2 = Time(1, 50)

print(t1 + t2)  
print(t1 < t2)      
```
